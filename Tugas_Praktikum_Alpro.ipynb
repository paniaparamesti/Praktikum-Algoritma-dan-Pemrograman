# **2. Luas lingkaran Dengan pi = 3,14**

```python
# Menghitung luas lingkaran dengan pi = 3.14
pi = 3.14
r = 7

luas_lingkaran = pi * r * r
print("Luas lingkaran:", luas_lingkaran)
```

 # **3. Total yang di butuhkan agung dan sandy untuk membayar**

```python
# Buku = b, pensil = s, penghapus = p
b = 3000
s = 1000
p = 500

# bayar = (x*b + x*s + x*p)
agung = 3*b + 2*s + 0*p
print("agung:", agung)

sandy = 2*b + 0*s + 4*p
print("sandy:", sandy)
```

# **1. Nilai Terbesar**

```python
x = 1
y = 49
z = 99

if x >= y and x >= z:
    terbesar = x
elif y >= x and y >= z:
    terbesar = y
else:
    terbesar = z

print("Nilai terbesar adalah", terbesar)
```
